# Contextualização

O Índice de Desenvolvimento Social é calculado pelo Insituto Pereira Passos -- instituto de pesquisa da prefeitura do Rio de Janeiro -- a fim de medir o grau de desenvolvimento social de uma determinada área geográfica em comparação com outras, as quais são, além do próprio município, recortes ou subdivisões intramunicipais, como setores censitários, bairros, favelas.

Oficialmente, o IPP só calcula o IDS para o município do Rio de Janeiro. Contudo, como o Índice baseia-se inteiramente em variáveis dos Censo Demográfico, é perfeitamente possível aplicar a sua metodologia para analisar o cenário de desenvolvimento social em outros municípios.

-----

## Objetivo Geral

Este pipeline toma como cenário hipotético a necessidade da administração municipal de Belo Horizonte, capital do estado de Minas Gerais, de realizar uma análise do desenvolvimento social da população com base na metodologia do IDS.

Logo, **o objetivo geral deste projeto é calcular o IDS para o município de Belo Horizonte (MG)** e para seus diferentes recortes geográficos intramunicipais, incluindo setores censitários, bairros e favelas, além do próprio município, possibilitando a análise do desenvolvimento social em diferentes níveis territoriais.

# Fonte dos Dados

Para viabilizar o cálculo do Índice de Desenvolvimento Social (IDS) no município de Belo Horizonte, utilizam-se os microdados e agregados oficiais do **Censo Demográfico 2022**, produzidos e disponibilizados pelo **Instituto Brasileiro de Geografia e Estatística (IBGE)**.

Os dados utilizados encontram-se agregados no menor nível territorial de divulgação pública do Censo: o **setor censitário**. Isso permite a análise tanto na escala de setor quanto a agregação hierárquica para bairros, favelas/comunidades urbanas e o município como um todo.


## 1. Origem e Conjuntos de Dados Obtidos

Todos os arquivos foram extraídos do repositório público de estatísticas do IBGE, na seção dedicada aos [Agregados por Setor Censitário em formato CSV](https://www.ibge.gov.br/estatisticas/downloads-estatisticas.html?caminho=Censos/Censo_Demografico_2022/Agregados_por_Setores_Censitarios/Agregados_por_Setor_csv/).

Os arquivos selecionados para o projeto contemplam as seguintes bases temáticas e metadados:

1. **`basico.csv` (Agregados por Setores Censitários - Básico):**
   - Fornece a malha cadastral e geográfica de referência, contendo a identificação do setor censitário (`CD_SETOR`), Unidade da Federação (`NM_UF`), Município (`NM_MUN`), Bairro (`NM_BAIRRO`) e classificação do setor em Favelas e Comunidades Urbanas (`NM_FCU`).
2. **`alfabetizacao.csv` (Agregados por Setores Censitários - Alfabetização):**
   - Contém o quantitativo de pessoas alfabetizadas e não alfabetizadas por faixas etárias, insumo essencial para a construção do indicador de taxa de analfabetismo (15 anos ou mais).
3. **`dom1.csv` (Agregados por Setores Censitários - Domicílios 1):**
   - Traz a quantidade de domicílios particulares permanentes ocupados e a contagem total de moradores por setor censitário.
4. **`dom2.csv` (Agregados por Setores Censitários - Domicílios 2):**
   - Fornece os dados de infraestrutura domiciliar urbana: ligação à rede geral de abastecimento de água, esgotamento sanitário (rede geral ou pluvial), cobertura de coleta de lixo e número de banheiros existentes no domicílio.
5. **`renda.csv` (Agregados por Setores Censitários - Renda):**
   - Contém as métricas de rendimento nominal médio mensal das pessoas responsáveis pelos domicílios com rendimento.
6. **Dicionários de Metadados (`dicionario_agregados.csv` e `dicionario_rendimentos.csv`):**
   - Contêm as descrições literais, conceitos e o mapeamento de cada código de variável (por exemplo, `V00111`, `V00309`, `V06004`), assegurando a correta catalogação e interpretação estatística dos campos ao longo do pipeline.

# 1. Ingestão (Camada Bronze)
Código que lê os CSVs do Volume e salva as tabelas brutas em formato Delta. 

```python
# Vou padronizar a leitura de todos os arquivos usando ponto e vírgula e encoding latin1 por conta da formatação do IBGE.
# As tabelas serão salvas no formato Delta dentro do esquema bronze para garantir o histórico bruto.

# 1. Arquivo Básico
df_basico = spark.read.format("csv") \
  .option("header", "true") \
  .option("sep", ";") \
  .option("inferSchema", "true") \
  .option("encoding", "latin1") \
  .load("/Volumes/IDS_MVP/bronze/dados_censo_2022/basico.csv")

df_basico.write.format("delta").mode("overwrite").saveAsTable("IDS_MVP.bronze.basico")

# 2. Arquivo de Alfabetização
df_alfabetizacao = spark.read.format("csv") \
  .option("header", "true") \
  .option("sep", ";") \
  .option("inferSchema", "true") \
  .option("encoding", "utf8") \
  .load("/Volumes/IDS_MVP/bronze/dados_censo_2022/alfabetizacao.csv")

df_alfabetizacao.write.format("delta").mode("overwrite").saveAsTable("IDS_MVP.bronze.alfabetizacao")

# 3. Arquivo Domicílios 1
df_dom1 = spark.read.format("csv") \
  .option("header", "true") \
  .option("sep", ";") \
  .option("inferSchema", "true") \
  .option("encoding", "utf8") \
  .load("/Volumes/IDS_MVP/bronze/dados_censo_2022/dom1.csv")

df_dom1.write.format("delta").mode("overwrite").saveAsTable("IDS_MVP.bronze.dom1")

# 4. Arquivo Domicílios 2
df_dom2 = spark.read.format("csv") \
  .option("header", "true") \
  .option("sep", ";") \
  .option("inferSchema", "true") \
  .option("encoding", "utf8") \
  .load("/Volumes/IDS_MVP/bronze/dados_censo_2022/dom2.csv")

df_dom2.write.format("delta").mode("overwrite").saveAsTable("IDS_MVP.bronze.dom2")

# 5. Arquivo de Renda
df_renda = spark.read.format("csv") \
  .option("header", "true") \
  .option("sep", ";") \
  .option("inferSchema", "true") \
  .option("encoding", "utf8") \
  .load("/Volumes/IDS_MVP/bronze/dados_censo_2022/renda.csv")

df_renda.write.format("delta").mode("overwrite").saveAsTable("IDS_MVP.bronze.renda")

# Ingestão do Dicionário Agregados
df_dic_basico = spark.read.format("csv") \
  .option("header", "true") \
  .option("sep", ";") \
  .option("inferSchema", "true") \
  .option("encoding", "utf8") \
  .load("/Volumes/ids_mvp/bronze/dados_censo_2022/dicionários/dicionario_agregados.csv")

df_dic_basico.write.format("delta").mode("overwrite").saveAsTable("IDS_MVP.bronze.dicionario_agregados")

# Ingestão do Dicionário Renda
df_dic_basico = spark.read.format("csv") \
  .option("header", "true") \
  .option("sep", ";") \
  .option("inferSchema", "true") \
  .option("encoding", "utf8") \
  .load("/Volumes/ids_mvp/bronze/dados_censo_2022/dicionários/dicionario_rendimentos.csv")

df_dic_basico.write.format("delta").mode("overwrite").saveAsTable("IDS_MVP.bronze.dicionario_renda")
```

Esses dados brutos, na forma como foram ingeridos, possuem todos os setores censitários do Brasil. ## 2.1. Filtrando os setores censitários
A fim de obter somente aqueles que são do município de Belo Horizonte (MG). Para isso, será necessário criar um dataframe que funcione como referência para identificar os setores, bairros e favelas (ou comunidades urbanas) do município de interesse. Os passos abaixos são requisitos para tornar esse objetivo exequível.

### 2.1.1. Padronizando o nome da chave: o código do Setor Censitário

Para tornar possível a consolidação da cesta de variáveis que irão compôr os indicadores do IDS, é preciso, antes, padornizar as variáveis que possuem os códigos do setores censitários. Depois de alguns testes, foi identificado que os dados publicados pelo IBGE a respeito dos resultados do Universo na forma de agregados por setores censitários apresentam diferentes nomes de variáveis:
1. Os dados do tema de Alfabetização apresentam a variável como "CD_setor";
2. Os dados do tema de Domicílios 2 apresentam a variável como "setor";

Padronizando o nome dessas variáveis, será possível dar continuidade aos objetivos elencados no início da seção 2.

```python
from pyspark.sql.functions import col

# lendo as tabelas brutas que ja foram ingeridas na bronze
df_basico = spark.table("IDS_MVP.bronze.basico")
df_alfabetizacao = spark.table("IDS_MVP.bronze.alfabetizacao")
df_dom1 = spark.table("IDS_MVP.bronze.dom1")
df_dom2 = spark.table("IDS_MVP.bronze.dom2")
df_renda = spark.table("IDS_MVP.bronze.renda")

# primeiro passo: padronizar a chave do setor
df_alfabetizacao = df_alfabetizacao.withColumnRenamed("CD_setor", "CD_SETOR")
df_dom2 = df_dom2.withColumnRenamed("setor", "CD_SETOR")
```

### 2.1.2. Criação a relação setor-bairro-favela para o município de Belo Horizonte (MG)

```python
# o basico tem o municipio e a uf de cada setor
# filtrando apenas Belo Horizonte / Minas Gerais
relacao_setor_bairro = (
    df_basico
    .filter(col("NM_UF") == "Minas Gerais")
    .filter(col("NM_MUN") == "Belo Horizonte")
    .select("CD_SETOR", "NM_UF", "NM_MUN", "NM_BAIRRO", "NM_FCU")
    .dropDuplicates(["CD_SETOR"])
)

# relacao setor-bairro e favelas do municipio
# fica de referencia para os joins e para filtrar as outras tabelas
relacao_setor_bairro.write.format("delta").mode("overwrite").saveAsTable("IDS_MVP.silver.relacao_setor_bairro")

# print de checagem no recorte ou subdivisao de setores censitarios
print("setores em BH:", relacao_setor_bairro.count())
relacao_setor_bairro.show(10, truncate=False)
```

### 2.1.3. Filtrando os setores dos resultados do Universo do Censo Demográfico de 2022 agregados para os setores censitários de Belo Horizonte (MG)


```python
# lista de setores que ficaram na relacao
# o volume de BH cabe nesse collect sem problema
setores_bh = [linha.CD_SETOR for linha in relacao_setor_bairro.select("CD_SETOR").collect()]

# cada tabela da bronze, menos os dicionarios, fica so com esses setores
df_basico_bh = df_basico.filter(col("CD_SETOR").isin(setores_bh))
df_alfabetizacao_bh = df_alfabetizacao.filter(col("CD_SETOR").isin(setores_bh))
df_dom1_bh = df_dom1.filter(col("CD_SETOR").isin(setores_bh))
df_dom2_bh = df_dom2.filter(col("CD_SETOR").isin(setores_bh))
df_renda_bh = df_renda.filter(col("CD_SETOR").isin(setores_bh))

# confere o tamanho depois do recorte
print("basico:", df_basico_bh.count())
print("alfabetizacao:", df_alfabetizacao_bh.count())
print("dom1:", df_dom1_bh.count())
print("dom2:", df_dom2_bh.count())
print("renda:", df_renda_bh.count())
```

No código e *outputs* acima, é possível verificar que, na camada Bronze, as tabelas possuíam uma quantidade de setores muito maior. Agora, na camada Silver, os dados foram filtrados e é importante destacar algo: o município de Belo Horizonte possui 5.166 setores censitários. Contudo, como alguns setores não possuem moradores, o IBGE não os insere nos dados temáticos dos agregados por setores censitários. Portanto, as tabelas de Renda, Alfabetização, Domicílios 1 e 2, possuem apenas 5.137 setores e esse é o esperado. Essa informação pode ser confirmada no link a seguir, no qual o IBGE informa que "O total de setores nos arquivos de Agregados por Setores Censtitários pode ser menor do que o total de setores existentes pois setores sem moradores não são incluídos no arquivo.": [Panorama do Censo](https://www.ibge.gov.br/estatisticas/sociais/populacao/22827-censo-demografico-2022.html?edicao=41852&t=resultados).

## 2.3. Verificação e remoção de valores anômalos

Todas as variáveis devem ser do tipo int, exceto a variável V06004 do tema de Renda, que deve ser float. Verificação abaixo:

```python
# printSchema mostra o tipo que o Spark esta usando agora
# string, int, bigint, double, float etc
#
# se a coluna ja veio como int/double, texto tipo "X" ou "-"
# nao esta mais no campo. na leitura com inferSchema esse
# texto vira null e some da checagem do passo 2.
# se a coluna ainda e string, o texto anomalo continua la

print("==== schema dom1 ====")
df_dom1_bh.printSchema()
print(df_dom1_bh.dtypes)

print("==== schema dom2 ====")
df_dom2_bh.printSchema()
print(df_dom2_bh.dtypes)

print("==== schema alfabetizacao ====")
df_alfabetizacao_bh.printSchema()
print(df_alfabetizacao_bh.dtypes)

print("==== schema renda ====")
df_renda_bh.printSchema()
print(df_renda_bh.dtypes)

print("==== schema basico ====")
df_basico_bh.printSchema()
print(df_basico_bh.dtypes)
```

Com isso, será preciso remover todos os valores X, substituindo-os por null e passar todas as variáveis para int, com exceção da V06004, que deve ser float. 

##2.4. Transformação das variáveis dos agregados por setores censitários em numéricas (int ou float);

```python
from pyspark.sql.functions import col, when, trim, regexp_replace

# X (com ou sem espaco na borda) vira nulo
# o restante permanece como esta ate o cast


# dom1
# -----------------
df_dom1_bh = (
    df_dom1_bh
    .withColumn("V00001", when(trim(col("V00001")) == "X", None).otherwise(col("V00001")))
    .withColumn("V00005", when(trim(col("V00005")) == "X", None).otherwise(col("V00005")))
    .withColumn("V00001", col("V00001").cast("int"))
    .withColumn("V00005", col("V00005").cast("int"))
)


# dom2
# -----------------
df_dom2_bh = (
    df_dom2_bh
    .withColumn("V00111", when(trim(col("V00111")) == "X", None).otherwise(col("V00111")))
    .withColumn("V00309", when(trim(col("V00309")) == "X", None).otherwise(col("V00309")))
    .withColumn("V00397", when(trim(col("V00397")) == "X", None).otherwise(col("V00397")))
    .withColumn("V00398", when(trim(col("V00398")) == "X", None).otherwise(col("V00398")))
    .withColumn("V00485", when(trim(col("V00485")) == "X", None).otherwise(col("V00485")))
    .withColumn("V00486", when(trim(col("V00486")) == "X", None).otherwise(col("V00486")))
    .withColumn("V00487", when(trim(col("V00487")) == "X", None).otherwise(col("V00487")))
    .withColumn("V00488", when(trim(col("V00488")) == "X", None).otherwise(col("V00488")))
    .withColumn("V00489", when(trim(col("V00489")) == "X", None).otherwise(col("V00489")))
    .withColumn("V00490", when(trim(col("V00490")) == "X", None).otherwise(col("V00490")))
    .withColumn("V00491", when(trim(col("V00491")) == "X", None).otherwise(col("V00491")))
    .withColumn("V00492", when(trim(col("V00492")) == "X", None).otherwise(col("V00492")))
    .withColumn("V00493", when(trim(col("V00493")) == "X", None).otherwise(col("V00493")))
    .withColumn("V00111", col("V00111").cast("int"))
    .withColumn("V00309", col("V00309").cast("int"))
    .withColumn("V00397", col("V00397").cast("int"))
    .withColumn("V00398", col("V00398").cast("int"))
    .withColumn("V00485", col("V00485").cast("int"))
    .withColumn("V00486", col("V00486").cast("int"))
    .withColumn("V00487", col("V00487").cast("int"))
    .withColumn("V00488", col("V00488").cast("int"))
    .withColumn("V00489", col("V00489").cast("int"))
    .withColumn("V00490", col("V00490").cast("int"))
    .withColumn("V00491", col("V00491").cast("int"))
    .withColumn("V00492", col("V00492").cast("int"))
    .withColumn("V00493", col("V00493").cast("int"))
)


# alfabetizacao
# -----------------
df_alfabetizacao_bh = (
    df_alfabetizacao_bh
    .withColumn("V00852", when(trim(col("V00852")) == "X", None).otherwise(col("V00852")))
    .withColumn("V00853", when(trim(col("V00853")) == "X", None).otherwise(col("V00853")))
    .withColumn("V00854", when(trim(col("V00854")) == "X", None).otherwise(col("V00854")))
    .withColumn("V00855", when(trim(col("V00855")) == "X", None).otherwise(col("V00855")))
    .withColumn("V00856", when(trim(col("V00856")) == "X", None).otherwise(col("V00856")))
    .withColumn("V00857", when(trim(col("V00857")) == "X", None).otherwise(col("V00857")))
    .withColumn("V00852", col("V00852").cast("int"))
    .withColumn("V00853", col("V00853").cast("int"))
    .withColumn("V00854", col("V00854").cast("int"))
    .withColumn("V00855", col("V00855").cast("int"))
    .withColumn("V00856", col("V00856").cast("int"))
    .withColumn("V00857", col("V00857").cast("int"))
)


# renda

# tira o X
# se o decimal vier com virgula, troca por ponto so no cast
df_renda_bh = (
    df_renda_bh
    .withColumn("V06004", when(trim(col("V06004")) == "X", None).otherwise(col("V06004")))
    .withColumn("V06004", regexp_replace(col("V06004"), ",", ".").cast("float"))
)
```

# 3. Modelagem (Camada Gold)


Nesta etapa, a camada Gold será construída a partir dos dados devidamente modelados. Essa modelagem consiste na construção dos 6 indicadores do IDS (calculados a partir das variáveis do Censo Demográfico de 2022 selecionadas na camada Silver). 

Basicamente, o passo-a-passo seguido será este:

3.1. Realizar um *join* (junção) da tabela relacao_setor_bairro, colocando-a à esquerda, com as tabelas  alfabetizacao, dom1, dom2, renda e basico;

3.2. Fazer uma tabela com esses mesmos dados agregados por bairro; fazer outra com eles agregados por favela (NM_FCU) e outra para todo o município. Isso totalizará 4 tabelas: 1) por setores censitários; 2) por bairros; 3) por favelas; 4) por município;

3.3. Feito isso, os indicadores do IDS serão calculados. A definição de cada um deles e suas respectivas fórmulas (baseadas nas variáveis do Censo 2022) são detalhadas na seção logo abaixo. Veja a seção "3.1. Descrição do Cálculo dos Indicadores do IDS";

3.4. Após o cálculo dos seis indicadores, determinam-se os valores máximos e mínimos para cada métrica considerando todos os níveis de agregação (setores, bairros, favelas e município). Esses limites formam uma tabela de referência unificada, que serve de base para a normalização dos dados e garante a comparabilidade dos resultados entre diferentes dimensões espaciais. Por fim, a normalização é aplicada a todos os indicadores, com exceção do Indicador 5 (Taxa de Analfabetismo), cuja escala é invertida em atendimento à diretriz metodológica. O cálculo de normalização e o da inversão encontra-se na seção "3.2. Descrição do Cálculo da Normalização e da Inversão de Indicadores";

3.5. Em seguida, somamos os valores obtidos em cada um dos **seis** indicadores normalizados para cada localidade e calculamos a sua média aritmética (dividindo a soma total por seis). Essa média corresponde ao valor final do Índice de Desenvolvimento Social (IDS) daquele local.

Vale notar que, por ser uma média calculada a partir de múltiplos indicadores normalizados, o IDS resultará em um valor contínuo na escala de 0 a 1, sem necessariamente atingir os extremos absolutos (0 ou 1). Ainda assim, o índice preserva a sua hierarquização ordinal, permitindo a correta diferenciação, ordenamento e comparação entre os setores censitários.

```python
%mod
Os indicadores são:

1. Forma adequada de abastecimento de água: Percentagem de domicílios particulares permanentes ocupados ligados à rede geral de distribuição sobre o total de domicílios particulares permanentes ocupados;

2. Esgotamento sanitário adequado: Percentagem de domicílios particulares permanentes ocupados com esgoto ligado à rede geral de esgoto ou à rede pluvial sobre o total de domicílios particulares permanentes ocupados;

3. Coleta de lixo adequada: Percentagem de domicílios particulares permanentes ocupados com lixo coletado diretamente por serviço de limpeza ou depositado em caçamba de serviço de limpeza sobre o total de domicílios particulares permanentes ocupados;

4. Habitabilidade (Banheiros por pessoa): Número médio de banheiros por pessoa, correspondente à razão entre o total de banheiros de uso exclusivo (com chuveiro e vaso sanitário) e o número total de moradores do domicílio;

5. Taxa de analfabetismo (15 anos ou mais): Percentagem de moradores de 15 anos e mais de idade que não sabem ler e escrever em relação ao total de moradores dessa mesma faixa etária;

6. Rendimento médio do responsável: Rendimento médio nominal mensal do responsável pelo domicílio particular permanente ocupado, excluindo-se da média os responsáveis declarados "sem rendimento".

OBS: para identificar as variáveis que os compõem, verificar a seção 2.2.
```

## 3.1. As cestas de variáveis

São consideradas cestas de variáveis o conjunto de variáveis que constituem o insumo para o cálculo do IDS. Essas cestas devem ser agregadas por todos os recortes intramunicipais analisados, isto é, setores censitários -- o próprio *join* completará essa etapa --, bairros, favelas e, finalmente, município.

Nas agregações feitas (*groupby*) pelas subdivisões territoriais listadas acima, todas as variáveis devem ser somadas, com exceção da V06004 (renda), que deve ser agregada por média. 


```python
from pyspark.sql.functions import col, sum as spark_sum, avg

# leio da silver o que ja esta limpo e tipado
relacao = spark.table("IDS_MVP.silver.relacao_setor_bairro")
dom1 = spark.table("IDS_MVP.silver.dom1")
dom2 = spark.table("IDS_MVP.silver.dom2")
alfa = spark.table("IDS_MVP.silver.alfabetizacao")
renda = spark.table("IDS_MVP.silver.renda")


# 3.1.1 join a esquerda
# a relacao fica a esquerda para nao perder setor
# setor sem morador entra com as V... nulas

cesta_setor = (
    relacao
    .join(dom1, on="CD_SETOR", how="left")
    .join(dom2, on="CD_SETOR", how="left")
    .join(alfa, on="CD_SETOR", how="left")
    .join(renda, on="CD_SETOR", how="left")
)


# 3.1.2 o que entra no groupBy
# quantidade (domicilio, morador, banheiro, alfabetizacao): soma
# V06004 (renda media ja calculada no setor): media

agregacoes = [
    spark_sum("V00001").alias("V00001"),
    spark_sum("V00005").alias("V00005"),
    spark_sum("V00111").alias("V00111"),
    spark_sum("V00309").alias("V00309"),
    spark_sum("V00397").alias("V00397"),
    spark_sum("V00398").alias("V00398"),
    spark_sum("V00485").alias("V00485"),
    spark_sum("V00486").alias("V00486"),
    spark_sum("V00487").alias("V00487"),
    spark_sum("V00488").alias("V00488"),
    spark_sum("V00489").alias("V00489"),
    spark_sum("V00490").alias("V00490"),
    spark_sum("V00491").alias("V00491"),
    spark_sum("V00492").alias("V00492"),
    spark_sum("V00493").alias("V00493"),
    spark_sum("V00852").alias("V00852"),
    spark_sum("V00853").alias("V00853"),
    spark_sum("V00854").alias("V00854"),
    spark_sum("V00855").alias("V00855"),
    spark_sum("V00856").alias("V00856"),
    spark_sum("V00857").alias("V00857"),
    avg("V06004").alias("V06004")
]

# bairro
cesta_bairro = (
    cesta_setor
    .groupBy("NM_UF", "NM_MUN", "NM_BAIRRO")
    .agg(*agregacoes)
)

# favela
# setor sem NM_FCU nao e favela, por isso o filtro
cesta_favela = (
    cesta_setor
    .filter(col("NM_FCU").isNotNull())
    .groupBy("NM_UF", "NM_MUN", "NM_FCU")
    .agg(*agregacoes)
)

# municipio
cesta_municipio = (
    cesta_setor
    .groupBy("NM_UF", "NM_MUN")
    .agg(*agregacoes)
)

# as quatro cestas no mesmo dicionario
# chave = subdivisao territorial
cestas = {
    "setor": cesta_setor,
    "bairro": cesta_bairro,
    "favela": cesta_favela,
    "municipio": cesta_municipio
}
```

## 3.2. Cálculo dos Indicadores do IDS


**1. Forma adequada de abastecimento de água**
Percentagem de domicílios particulares permanentes ocupados ligados à rede geral de distribuição sobre o total de domicílios particulares permanentes ocupados.
$$ Indicador\_1 = \left( \frac{V00111}{V00001} \right) \times 100 $$

**2. Esgotamento sanitário adequado**
Percentagem de domicílios particulares permanentes ocupados com esgoto ligado à rede geral de esgoto ou à rede pluvial sobre o total de domicílios particulares permanentes ocupados.
$$ Indicador\_2 = \left( \frac{V00309}{V00001} \right) \times 100 $$

**3. Coleta de lixo adequada**
Percentagem de domicílios particulares permanentes ocupados com lixo coletado diretamente por serviço de limpeza ou depositado em caçamba de serviço de limpeza sobre o total de domicílios particulares permanentes ocupados.
$$ Indicador\_3 = \left( \frac{V00397 + V00398}{V00001} \right) \times 100 $$

**4. Habitabilidade (Banheiros por pessoa)**
Número médio de banheiros por pessoa, correspondente à razão entre o total de banheiros de uso exclusivo (com chuveiro e vaso sanitário) e o número total de moradores do domicílio.
$$ Indicador\_4 = \frac{V00485 + (V00486 \times 2) + \dots + (V00493 \times 9)}{V00005} $$
*Nota: O numerador soma a quantidade total de banheiros, multiplicando o número de domicílios pela quantidade de banheiros que possuem (de 1 até 9 ou mais).*

**5. Taxa de analfabetismo (15 anos ou mais)**
Percentagem de moradores de 15 anos e mais de idade que não sabem ler e escrever em relação ao total de moradores dessa mesma faixa etária.
$$ Indicador\_5 = \left( \frac{V00853 + V00855 + V00857}{V00852 + V00853 + V00854 + V00855 + V00856 + V00857} \right) \times 100 $$

**6. Rendimento médio do responsável**
Rendimento médio nominal mensal do responsável pelo domicílio particular permanente ocupado, excluindo-se da média os responsáveis declarados "sem rendimento", expresso em salários mínimos (o salário mínimo para 2022 foi de R$1212,00. [Veja o texto integral da Lei nº 14.358, de 1º de junho de 2022](https://www.planalto.gov.br/ccivil_03/_ato2019-2022/2022/lei/l14358.htm)).
$$ Indicador\_6 = V06004/1212 $$

**OBS: FAZ PARTE DA METODOLOGIA DO IDS SUBSTITUIR VALORES NULOS POR 0, PARA QUE ALGUNS SETORES QUE POSSUAM INFORMAÇÃO SOBRE ALGUNS INDICADORES NÃO TENHAM SEU IDS ANULADO EM RAZÃO DA AUSÊNCIA DE OUTROS INDICADORES**. 

```python
from pyspark.sql.functions import col, sum as spark_sum, avg, when, coalesce, lit

# salario minimo de 2022
sm_2022 = 1212.0

# leio da silver
relacao = spark.table("IDS_MVP.silver.relacao_setor_bairro")
dom1 = spark.table("IDS_MVP.silver.dom1")
dom2 = spark.table("IDS_MVP.silver.dom2")
alfa = spark.table("IDS_MVP.silver.alfabetizacao")
renda = spark.table("IDS_MVP.silver.renda")

# join a esquerda: a relacao segura todos os setores
cesta_setor = (
    relacao
    .join(dom1, on="CD_SETOR", how="left")
    .join(dom2, on="CD_SETOR", how="left")
    .join(alfa, on="CD_SETOR", how="left")
    .join(renda, on="CD_SETOR", how="left")
)

# quantidades somam
# V06004: media PRIMEIRO (so nas subdivisoes acima do setor censitario)
agregacoes = [
    spark_sum("V00001").alias("V00001"),
    spark_sum("V00005").alias("V00005"),
    spark_sum("V00111").alias("V00111"),
    spark_sum("V00309").alias("V00309"),
    spark_sum("V00397").alias("V00397"),
    spark_sum("V00398").alias("V00398"),
    spark_sum("V00485").alias("V00485"),
    spark_sum("V00486").alias("V00486"),
    spark_sum("V00487").alias("V00487"),
    spark_sum("V00488").alias("V00488"),
    spark_sum("V00489").alias("V00489"),
    spark_sum("V00490").alias("V00490"),
    spark_sum("V00491").alias("V00491"),
    spark_sum("V00492").alias("V00492"),
    spark_sum("V00493").alias("V00493"),
    spark_sum("V00852").alias("V00852"),
    spark_sum("V00853").alias("V00853"),
    spark_sum("V00854").alias("V00854"),
    spark_sum("V00855").alias("V00855"),
    spark_sum("V00856").alias("V00856"),
    spark_sum("V00857").alias("V00857"),
    avg("V06004").alias("V06004")
]

cesta_bairro = (
    cesta_setor
    .groupBy("NM_UF", "NM_MUN", "NM_BAIRRO")
    .agg(*agregacoes)
)

cesta_favela = (
    cesta_setor
    .filter(col("NM_FCU").isNotNull())
    .groupBy("NM_UF", "NM_MUN", "NM_FCU")
    .agg(*agregacoes)
)

cesta_municipio = (
    cesta_setor
    .groupBy("NM_UF", "NM_MUN")
    .agg(*agregacoes)
)

# cestas: so variaveis do Censo + recorte
cestas = {
    "setor": cesta_setor,
    "bairro": cesta_bairro,
    "favela": cesta_favela,
    "municipio": cesta_municipio
}

def calcula_indicadores(df):
    # Substitui valores nulos por zero em todas as colunas numéricas antes das operações matemáticas
    df = df.fillna(0)
    
    # indicador 1: % agua na rede
    df = df.withColumn(
        "indicador_1",
        when(col("V00001") > 0, (col("V00111") / col("V00001")) * 100)
    )

    # indicador 2: % esgoto na rede
    df = df.withColumn(
        "indicador_2",
        when(col("V00001") > 0, (col("V00309") / col("V00001")) * 100)
    )

    # indicador 3: % lixo com coleta adequada
    df = df.withColumn(
        "indicador_3",
        when(col("V00001") > 0, ((col("V00397") + col("V00398")) / col("V00001")) * 100)
    )

    # indicador 4: banheiros por morador
    total_banheiros = (
        coalesce(col("V00485"), lit(0)) * 1 +
        coalesce(col("V00486"), lit(0)) * 2 +
        coalesce(col("V00487"), lit(0)) * 3 +
        coalesce(col("V00488"), lit(0)) * 4 +
        coalesce(col("V00489"), lit(0)) * 5 +
        coalesce(col("V00490"), lit(0)) * 6 +
        coalesce(col("V00491"), lit(0)) * 7 +
        coalesce(col("V00492"), lit(0)) * 8 +
        coalesce(col("V00493"), lit(0)) * 9
    )
    df = df.withColumn(
        "indicador_4",
        when(col("V00005") > 0, total_banheiros / col("V00005"))
    )

    # indicador 5: taxa de analfabetismo (15+)
    nao_le = coalesce(col("V00853"), lit(0)) + coalesce(col("V00855"), lit(0)) + coalesce(col("V00857"), lit(0))
    pop_15mais = (
        coalesce(col("V00852"), lit(0)) +
        coalesce(col("V00853"), lit(0)) +
        coalesce(col("V00854"), lit(0)) +
        coalesce(col("V00855"), lit(0)) +
        coalesce(col("V00856"), lit(0)) +
        coalesce(col("V00857"), lit(0))
    )
    df = df.withColumn(
        "indicador_5",
        when(pop_15mais > 0, (nao_le / pop_15mais) * 100)
    )

    # indicador 6: media da V06004 / salario minimo
    df = df.withColumn(
        "indicador_6",
        col("V06004") / lit(sm_2022)
    )

    return df

# calcula em cima da cesta
# indicadores: subdivisoes intramunicipais + os 6 indicadores
indicadores = {
    "setor": calcula_indicadores(cestas["setor"]).select(
        "CD_SETOR", "NM_UF", "NM_MUN", "NM_BAIRRO", "NM_FCU",
        "indicador_1", "indicador_2", "indicador_3",
        "indicador_4", "indicador_5", "indicador_6"
    ),
    "bairro": calcula_indicadores(cestas["bairro"]).select(
        "NM_UF", "NM_MUN", "NM_BAIRRO",
        "indicador_1", "indicador_2", "indicador_3",
        "indicador_4", "indicador_5", "indicador_6"
    ),
    "favela": calcula_indicadores(cestas["favela"]).select(
        "NM_UF", "NM_MUN", "NM_FCU",
        "indicador_1", "indicador_2", "indicador_3",
        "indicador_4", "indicador_5", "indicador_6"
    ),
    "municipio": calcula_indicadores(cestas["municipio"]).select(
        "NM_UF", "NM_MUN",
        "indicador_1", "indicador_2", "indicador_3",
        "indicador_4", "indicador_5", "indicador_6"
    )
}
```

## 3.3. Encontrando os valores mínimos e máximos.

Para cada uma das subdivisões territoriais (ou intramunicipais, como queira), serão extraídos os valores máximos e mínimos, para cada um dos indicadores. Esses valores irão compôr uma tabela de referência que será utilizada, na próxima seção, para a normalização dos indicadores.

```python
from pyspark.sql.functions import col, lit, min as spark_min, max as spark_max


# 1) empilha os 4 recortes no padrao da tabela de referencia
#    nivel, id_recorte, nm_recorte + indicadores


ind_setor = (
    indicadores["setor"]
    .select(
        lit("setor").alias("nivel"),
        col("CD_SETOR").cast("string").alias("id_recorte"),
        col("CD_SETOR").cast("string").alias("nm_recorte"),
        "indicador_1", "indicador_2", "indicador_3",
        "indicador_4", "indicador_5", "indicador_6"
    )
)

ind_bairro = (
    indicadores["bairro"]
    .select(
        lit("bairro").alias("nivel"),
        col("NM_BAIRRO").cast("string").alias("id_recorte"),
        col("NM_BAIRRO").cast("string").alias("nm_recorte"),
        "indicador_1", "indicador_2", "indicador_3",
        "indicador_4", "indicador_5", "indicador_6"
    )
)

ind_favela = (
    indicadores["favela"]
    .select(
        lit("favela").alias("nivel"),
        col("NM_FCU").cast("string").alias("id_recorte"),
        col("NM_FCU").cast("string").alias("nm_recorte"),
        "indicador_1", "indicador_2", "indicador_3",
        "indicador_4", "indicador_5", "indicador_6"
    )
)

ind_municipio = (
    indicadores["municipio"]
    .select(
        lit("municipio").alias("nivel"),
        col("NM_MUN").cast("string").alias("id_recorte"),
        col("NM_MUN").cast("string").alias("nm_recorte"),
        "indicador_1", "indicador_2", "indicador_3",
        "indicador_4", "indicador_5", "indicador_6"
    )
)


indicadores_unificado = (
    ind_setor
    .unionByName(ind_bairro)
    .unionByName(ind_favela)
    .unionByName(ind_municipio)
)


# 2) min e max de cada indicador, por nivel

ref_minmax = (
    indicadores_unificado
    .groupBy("nivel")
    .agg(
        spark_min("indicador_1").alias("min_indicador_1"),
        spark_max("indicador_1").alias("max_indicador_1"),
        spark_min("indicador_2").alias("min_indicador_2"),
        spark_max("indicador_2").alias("max_indicador_2"),
        spark_min("indicador_3").alias("min_indicador_3"),
        spark_max("indicador_3").alias("max_indicador_3"),
        spark_min("indicador_4").alias("min_indicador_4"),
        spark_max("indicador_4").alias("max_indicador_4"),
        spark_min("indicador_5").alias("min_indicador_5"),
        spark_max("indicador_5").alias("max_indicador_5"),
        spark_min("indicador_6").alias("min_indicador_6"),
        spark_max("indicador_6").alias("max_indicador_6")
    )
)

print("unificado:", indicadores_unificado.count())
indicadores_unificado.show(5, truncate=False)
ref_minmax.show(truncate=False)
```

## 3.4. Normalizando indicadores e calculando o IDS

Inicialmente, os indicadores 1, 2, 3, 4 e 6 serão normalizados conforme a fórmula abaixo, utilizando-se dos valores máximos e mínimos encontrados anteriormente.

$$VN_{ij} = 1 - \frac{MV_i - V_{ij}}{MV_i - mV_i}$$

VNij = valor normalizado na escala de 0 a 1 do indicador i no lugar j; \
MVi = maior valor obtido pelo indicador i entre todos os recortes geográficos pesquisados; \
mVi = menor valor obtido pelo indicador i entre todos os recortes geográficos pesquisados; \
Vij = valor obtido pelo indicador i no lugar j.

Contudo, para o cálculo do indicador 5, a fórmula da normalização será invertida, conforme destacado na introdução desta seção:

$$VN_{ij} = 1 - \frac{mV_i - V_{ij}}{mV_i - MV_i}$$


```python
from pyspark.sql.functions import col, lit, min as spark_min, max as spark_max

# min e max de cada indicador em TODOS os recortes juntos
# setor, bairro, favela e municipio entram na mesma conta
limites = indicadores_unificado.agg(
    spark_min("indicador_1").alias("min_1"),
    spark_max("indicador_1").alias("max_1"),
    spark_min("indicador_2").alias("min_2"),
    spark_max("indicador_2").alias("max_2"),
    spark_min("indicador_3").alias("min_3"),
    spark_max("indicador_3").alias("max_3"),
    spark_min("indicador_4").alias("min_4"),
    spark_max("indicador_4").alias("max_4"),
    spark_min("indicador_5").alias("min_5"),
    spark_max("indicador_5").alias("max_5"),
    spark_min("indicador_6").alias("min_6"),
    spark_max("indicador_6").alias("max_6")
).collect()[0]

min_1, max_1 = limites["min_1"], limites["max_1"]
min_2, max_2 = limites["min_2"], limites["max_2"]
min_3, max_3 = limites["min_3"], limites["max_3"]
min_4, max_4 = limites["min_4"], limites["max_4"]
min_5, max_5 = limites["min_5"], limites["max_5"]
min_6, max_6 = limites["min_6"], limites["max_6"]

print("limites globais")
print("1:", min_1, max_1)
print("2:", min_2, max_2)
print("3:", min_3, max_3)
print("4:", min_4, max_4)
print("5:", min_5, max_5)
print("6:", min_6, max_6)


# normalizacao
# 1,2,3,4,6: VN = 1 - (max - V) / (max - min)
# 5:         VN = 1 - (min - V) / (min - max)
#            analfabetismo alto vira nota baixa

def normaliza_e_ids(df):
    df = df.withColumn(
        "indicador_1_norm",
        lit(1) - (lit(max_1) - col("indicador_1")) / lit(max_1 - min_1)
    )
    df = df.withColumn(
        "indicador_2_norm",
        lit(1) - (lit(max_2) - col("indicador_2")) / lit(max_2 - min_2)
    )
    df = df.withColumn(
        "indicador_3_norm",
        lit(1) - (lit(max_3) - col("indicador_3")) / lit(max_3 - min_3)
    )
    df = df.withColumn(
        "indicador_4_norm",
        lit(1) - (lit(max_4) - col("indicador_4")) / lit(max_4 - min_4)
    )
    df = df.withColumn(
        "indicador_6_norm",
        lit(1) - (lit(max_6) - col("indicador_6")) / lit(max_6 - min_6)
    )

    # indicador 5 invertido
    df = df.withColumn(
        "indicador_5_norm",
        lit(1) - (lit(min_5) - col("indicador_5")) / lit(min_5 - max_5)
    )

    # IDS: media aritmetica dos 6 normalizados
    df = df.withColumn(
        "ids",
        (
            col("indicador_1_norm") +
            col("indicador_2_norm") +
            col("indicador_3_norm") +
            col("indicador_4_norm") +
            col("indicador_5_norm") +
            col("indicador_6_norm")
        ) / lit(6.0)
    )
    return df

indicadores["setor"] = normaliza_e_ids(indicadores["setor"])
indicadores["bairro"] = normaliza_e_ids(indicadores["bairro"])
indicadores["favela"] = normaliza_e_ids(indicadores["favela"])
indicadores["municipio"] = normaliza_e_ids(indicadores["municipio"])
```

### 1. Quais são os 10 bairros com os maiores Índices de Desenvolvimento Social (IDS) em Belo Horizonte e a respectiva renda deles de acordo com o indicador 6 não-normalizado? Com base nessa tabela, analise a homogeneidade ou heterogeneidade desses bairros de acordo com a renda média dos bairros. 

```python

display(spark.sql("""
    SELECT
        i.NM_BAIRRO,
        ROUND(i.ids, 4) AS ids,
        ROUND(i.indicador_6, 2) AS renda_sm,
        ROUND(c.V06004, 2) AS renda_reais
    FROM IDS_MVP.gold.indicadores_bairro AS i
    INNER JOIN IDS_MVP.gold.cesta_variaveis_bairro AS c
        ON  i.NM_UF     = c.NM_UF
        AND i.NM_MUN    = c.NM_MUN
        AND i.NM_BAIRRO = c.NM_BAIRRO
    WHERE i.ids IS NOT NULL
      AND i.NM_BAIRRO IS NOT NULL
    ORDER BY i.ids DESC
    LIMIT 10
"""))
```

Com base na tabela acima, é possível identificar que os bairros com os 10 maiores valores de IDS possuem, pelo menos, 10 salários mínimos (SM) em termos aproximados. Os valores máximos e mínmos correspondem, respectivamente, a 0.836 no bairro de Mangabeiras e aproximadamente 0.789 no bairro Luxemburgo.

Embora a tabela esteja ordenada pela variável IDS, do maior valor ao menor, é possível identificar que a variável renda não apresenta um comportamento de crescimento constante em todos os casos. Por exemplo, embora o bairro Sion apresente o 9º valor de IDS, a renda média é menor ao nível imediatamente inferior do IDS de acordo com os bairros: Luxemburgo, que está em 10º lugar, com 11.01 SM, possui um rendimento médio superior a Sion, com 10.21 SM, embora esse último esteja em 9º lugar.

### 4. Alocando o IDS das favelas ou comunidades urbanas em faixas categóricas, analise a concentração percentual dos registros nessas faixas.

```python
# binning do ids das favelas
# cada NM_FCU cai em uma faixa de 0.20
# a pergunta e concentracao percentual, nao o ranking individual

display(spark.sql("""
    SELECT
        faixa,
        COUNT(*) AS n_favelas,
        ROUND(100.0 * COUNT(*) / SUM(COUNT(*)) OVER (), 2) AS porcentagem
    FROM (
        SELECT
            CASE
                WHEN ids < 0.20 THEN '0.00–0.20 (muito baixo)'
                WHEN ids < 0.40 THEN '0.20–0.40 (baixo)'
                WHEN ids < 0.60 THEN '0.40–0.60 (medio)'
                WHEN ids < 0.80 THEN '0.60–0.80 (alto)'
                ELSE '0.80–1.00 (muito alto)'
            END AS faixa
        FROM IDS_MVP.gold.indicadores_favela
    )
    GROUP BY faixa
    ORDER BY faixa
"""))
```

A distribuição das 220 favelas e comunidades urbanas de Belo Horizonte demonstra uma acentuada concentração na faixa de IDS considerado alto (0.60 a 0.80), abrangendo 179 comunidades, o que representa expressivos 81.36% do total de registros analisados. Na sequência, a faixa de nível médio (0.40 a 0.60) reúne 38 territórios, correspondendo a 17.27% da amostra, enquanto a faixa de nível baixo (0.20 a 0.40) compreende apenas 3 localidades, perfazendo residuais 1.36%. Nenhuma comunidade foi classificada nas faixas extremas de muito baixo (inferior a 0.20) ou muito alto (superior a 0.80). 

Esse comportamento revela que, longe de apresentarem uma dispersão caótica ao longo de toda a régua de desenvolvimento social, as favelas da capital exibem uma notável convergência em torno do patamar mediano de 0.6 observado anteriormente. A quase total ausência de registros na faixa inferior a 0.40 e a inexistência de registros abaixo de 0.20 confirmam que a infraestrutura urbana básica e os serviços essenciais alcançam a esmagadora maioria desses assentamentos, estabelecendo um piso mínimo de qualidade de vida que evita situações generalizadas de indigência estrutural.

É importante destacar que não existem favelas cujo IDS esteja situado na faixa "muito alto". Contudo, esse cenário pode se mostrar diferente quando os setores censitários forem analisados.

### 5. Dentre a distribuição do IDS das favelas ou comunidades urbanas, quais estão abaixo do limite inferior?

```python
# q1, q3, iqr e a cerca inferior da distribuicao das favelas

stats_favela = spark.sql("""
    SELECT
        percentile(ids, 0.25) AS q1,
        percentile(ids, 0.75) AS q3,
        percentile(ids, 0.75) - percentile(ids, 0.25) AS iqr,
        percentile(ids, 0.25) - 1.5 * (percentile(ids, 0.75) - percentile(ids, 0.25)) AS limite_inferior
    FROM IDS_MVP.gold.indicadores_favela
""")

display(stats_favela)
```

Com o resultado acima, obtive-se o resultado necessário para identificar quais favelas possuem o IDS igual ou menor ao limite inferior encontrado:

```python
from pyspark.sql.functions import col
# favelas com ids <= limite inferior

display(
    spark.table("IDS_MVP.gold.indicadores_favela")
    .crossJoin(stats_favela)
    .filter(col("ids") <= col("limite_inferior"))
    .select("NM_FCU", "ids")
    .orderBy("ids")
)
```


São 37 favelas e comunidades urbanas de Belo Horizonte apresentam Índice de Desenvolvimento Social (IDS) igual ou inferior ao limite inferior estatístico calculado (0,5895), correspondendo a 16,82% do total das 220 comunidades mapeadas no município.

Dessa forma, essas 37 localidades configuram discrepantes estatísticos à esquerda (outliers negativos), afastando-se expressivamente da faixa central consolidada na qual se posiciona a maioria das comunidades da capital (como visto anteriormente, de 80% das favelas estão situadas na faixa que compreende um IDS de 0.6 a 0.8). 

A leitura integral desse contingente revela uma clara gradação interna de vulnerabilidade. No ponto mais crítico da distribuição situam-se as três únicas comunidades classificadas no estrato de nível baixo (IDS inferior a 0,40): Vitória (0,3465), Fidel Castro (0,3535) e Vila Esperança (0,3576). Em seguida, desponta um segmento intermediário de severidade composto por sete localidades com índices situados entre 0,40 e 0,50, abrangendo Vila Chaves (0,4047), Esperança (0,4065), Novo Lajedo (0,4111), Montes Claros (0,4339) etc. 

O contraste entre a menor pontuação observada (Vitória, com 0,3465) e a localidade de maior valor entre os discrepantes (Vila Canto do Sabiá, com 0,5807) resulta em uma amplitude interna de 0,2341 pontos, evidenciando que as privações estruturais não se distribuem de modo homogêneo nem mesmo entre as periferias mais fragilizadas. Enquanto 81,36% das comunidades de Belo Horizonte convergem para a faixa de IDS alto (0,60 a 0,80), este grupo de 37 territórios representa bolsões onde as carências socioeconômicas e urbanas persistem de forma aguda, elegendo-os como foco prioritário para eventuais intervenções por meio de políticas públicas. 


### 6. Existem setores censitários em situação de vulnerabilidade crítica (IDS < 0,50) situados dentro dos 10 bairros de maior IDS?

```python

display(spark.sql("""
    WITH top10 AS (
        SELECT NM_UF, NM_MUN, NM_BAIRRO, ids AS ids_bairro
        FROM IDS_MVP.gold.indicadores_bairro
        ORDER BY ids DESC
        LIMIT 10
    )
    SELECT
        t.NM_BAIRRO,
        ROUND(t.ids_bairro, 4) AS ids_bairro,
        s.CD_SETOR,
        s.NM_FCU,
        ROUND(s.ids, 4) AS ids_setor
    FROM IDS_MVP.gold.indicadores_setor AS s
    INNER JOIN top10 AS t
        ON  s.NM_UF     = t.NM_UF
        AND s.NM_MUN    = t.NM_MUN
        AND s.NM_BAIRRO = t.NM_BAIRRO
    WHERE s.ids <= 0.50
    ORDER BY s.ids ASC
"""))
```

### 7. Considerando as mesmas faixas categóricas de IDS, qual é a distribuição percentual dos setores censitários situados na malha urbana regular em comparação com os de favelas e comunidades urbanas?

```python

display(spark.sql("""
    WITH base AS (
        SELECT
            CASE WHEN NM_FCU IS NULL THEN 'urbano' ELSE 'favela' END AS tipo,
            CASE
                WHEN ids < 0.20 THEN '0.00–0.20 (muito baixo)'
                WHEN ids < 0.40 THEN '0.20–0.40 (baixo)'
                WHEN ids < 0.60 THEN '0.40–0.60 (medio)'
                WHEN ids < 0.80 THEN '0.60–0.80 (alto)'
                ELSE '0.80–1.00 (muito alto)'
            END AS faixa
        FROM IDS_MVP.gold.indicadores_setor
    ),
    totais AS (
        SELECT
            SUM(CASE WHEN tipo = 'urbano' THEN 1 ELSE 0 END) AS n_urbano,
            SUM(CASE WHEN tipo = 'favela' THEN 1 ELSE 0 END) AS n_favela
        FROM base
    )
    SELECT
        b.faixa,
        ROUND(100.0 * SUM(CASE WHEN b.tipo = 'urbano' THEN 1 ELSE 0 END) / MAX(t.n_urbano), 2) AS pct_area_urbanizada,
        ROUND(100.0 * SUM(CASE WHEN b.tipo = 'favela' THEN 1 ELSE 0 END) / MAX(t.n_favela), 2) AS pct_favela
    FROM base AS b
    CROSS JOIN totais AS t
    GROUP BY b.faixa
    ORDER BY b.faixa
"""))
```

A distribuição dos setores censitários revela que a expressiva maioria do território de Belo Horizonte converge para a faixa de IDS alto (0,60 a 0,80), patamar que abriga 95,11% dos setores da malha urbana regular e 85,23% dos setores situados em favelas e comunidades urbanas. Notavelmente, a faixa de IDS muito baixo (0,00 a 0,20) não registrou nenhuma ocorrência em qualquer uma das tipologias territoriais.

A clivagem socioespacial entre as duas realidades manifesta-se com vigor nas faixas de menor desenvolvimento social (abaixo de 0,60). Nas favelas, os estratos médios (12,36%) e baixo (2,27%) somam 14,63% dos setores, proporção quase 14 vezes superior à verificada na área urbanizada regular, que acumula apenas 1,06% nessa situação de vulnerabilidade (0,99% em nível médio e residuais 0,07% em nível baixo).
